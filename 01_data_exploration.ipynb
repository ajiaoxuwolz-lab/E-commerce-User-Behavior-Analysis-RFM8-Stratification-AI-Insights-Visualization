```python
import pandas as pd

# 1. 定义列名（因为原始数据无表头）
col_names = ["user_id", "item_id", "category_id", "behavior", "timestamp"]

# 2. 读取前100万行，指定列名和数据类型（内存优化）
df = pd.read_csv(
    "raw_data/UserBehavior.csv",
    nrows=1000000,
    names=col_names,
    dtype={
        "user_id": "int32",
        "item_id": "int32",
        "category_id": "int32",
        "behavior": "category"
    }
)

print("✅ 数据读取成功，数据集行列数：", df.shape)
print("\n前5行预览：")
print(df.head())

# 3. 可选：内存占用探查（加分项）
total_memory_mb = df.memory_usage(deep=True).sum() / 1024 ** 2
print(f"\n📊 数据集总内存占用：{total_memory_mb:.2f} MB")

# 4. 可选：时间戳范围探查（仅查看，不做转换）
print(f"\n⏰ timestamp 最小值：{df['timestamp'].min()}")
print(f"⏰ timestamp 最大值：{df['timestamp'].max()}")
```
